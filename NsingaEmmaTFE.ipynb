# **Analyse des composantes principales - ACP**

**Préparation des données**

On commence par normaliser les variables à forte corrélation.

```python
from sklearn.preprocessing import StandardScaler

# Sélection des variables pour l'ACP
bill_vars = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']
pay_vars = ['PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']

# Normalisation des variables
scaler = StandardScaler()
scaled_bills = scaler.fit_transform(df[bill_vars])
scaled_pays = scaler.fit_transform(df[pay_vars])
```

**Application de l'ACP**

On applique l'ACP séparément sur les montants des factures et les statuts de paiement.

```python
from sklearn.decomposition import PCA

# ACP sur les montants des factures
pca_bills = PCA()
pca_bills_fit = pca_bills.fit(scaled_bills)
bill_pcs = pca_bills.transform(scaled_bills)

# ACP sur les statuts de paiement
pca_pays = PCA()
pca_pays_fit = pca_pays.fit(scaled_pays)
pay_pcs = pca_pays.transform(scaled_pays)
```

**Explication de la variance**

Analysons la variance expliquée pour déterminer le nombre de composantes à conserver.

```python
# Variance expliquée pour les montants des factures
explained_variance_bills = pca_bills.explained_variance_ratio_
print("Variance expliquée (Factures) :", explained_variance_bills)

# Variance expliquée pour les statuts de paiement
explained_variance_pays = pca_pays.explained_variance_ratio_
print("Variance expliquée (Paiements) :", explained_variance_pays)
```

**Choix du nombre de composantes**

On utilise le scree plot pour visualiser la décroissance de la variance expliquée.

```python
# Scree plot pour les montants des factures
plt.figure(figsize=(10,6))
plt.plot(range(1, len(explained_variance_bills)+1), explained_variance_bills, marker='o', linestyle='--')
plt.title('Scree Plot - Factures')
plt.xlabel('Composante Principale')
plt.ylabel('Variance expliquée')
plt.show()

# Scree plot pour les statuts de paiement
plt.figure(figsize=(10,6))
plt.plot(range(1, len(explained_variance_pays)+1), explained_variance_pays, marker='o', linestyle='--')
plt.title('Scree Plot - Statuts de Paiement')
plt.xlabel('Composante Principale')
plt.ylabel('Variance expliquée')
plt.show()
```

Interprétation des Scree Plots : Choix du Nombre de Composantes
1. Scree Plot - Factures (BILL_AMT1 à BILL_AMT6)
Coude à la 2ème composante : La première composante principale explique plus de 80% de la variance.
Les composantes suivantes apportent très peu d'information supplémentaire.
Décision : On conserve 2 composantes principales pour les factures :
PC_BILL1 : Représente la tendance générale des montants des factures.
PC_BILL2 : Capture de légères variations (ex : fluctuations mensuelles).
2. Scree Plot - Statuts de Paiement (PAY_0 à PAY_6)
Coude à la 2ème composante : La première composante principale explique 70% de la variance.
La deuxième composante apporte une petite nuance aux comportements de paiement.
Décision : On conserve 2 composantes principales pour les statuts de paiement :
PC_PAY1 : Représente le comportement global de paiement.
PC_PAY2 : Capture les écarts ponctuels dans les retards de paiement.

Résumé des décisions :
2 composantes principales pour les montants des factures (PC_BILL1, PC_BILL2).
2 composantes principales pour les statuts de paiement (PC_PAY1, PC_PAY2).
Cela nous donne 4 nouvelles features non corrélées pour K-means et SVM.

**Projection des données**

On utilise les composantes principales sélectionnées pour créer de nouvelles features.

```python
# On garde les 2 premières composantes pour les factures
n_components_bills = 2
bills_pc = pca_bills.transform(scaled_bills)[:, :n_components_bills]

# On garde les 2 premières composantes pour les statuts de paiement
n_components_pays = 2
pays_pc = pca_pays.transform(scaled_pays)[:, :n_components_pays]

# Création d'un nouveau DataFrame avec les composantes principales
df_pca = df.copy()
df_pca['PC_BILL1'] = bills_pc[:, 0]
df_pca['PC_BILL2'] = bills_pc[:, 1]
df_pca['PC_PAY1'] = pays_pc[:, 0]
df_pca['PC_PAY2'] = pays_pc[:, 1]
```

# **Application de K-means avec les Composantes Principales**

**Choisir le nombre de clusters (Méthode du coude)**

On va utiliser l'Inertie intra-cluster pour déterminer le nombre optimal de clusters.

**Objectif** : Trouver le point de coude où l'ajout de clusters n'améliore plus significativement l'inertie.

**Décision** : On choisira le nombre optimal de clusters en fonction du graphique.

```python
from sklearn.cluster import KMeans
import numpy as np

# Préparation des données pour K-means
X = df_pca[['PC_BILL1', 'PC_BILL2', 'PC_PAY1', 'PC_PAY2']]

# Méthode du coude
inertia = []
range_clusters = range(1, 11)

for k in range_clusters:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

# Affichage du coude
plt.figure(figsize=(10,6))
plt.plot(range_clusters, inertia, marker='o', linestyle='--')
plt.title('Méthode du coude')
plt.xlabel('Nombre de clusters (k)')
plt.ylabel('Inertie intra-cluster')
plt.show()
```

### **Interprétation du Graphique de la Méthode du Coude**  

---

### **1. Observation du coude**  
- Le coude se situe clairement à **k = 3** et **k = 4**.  
- Après **k = 4**, l'amélioration de l'inertie devient **beaucoup moins significative**.  
- Cela signifie qu'à partir de 4 clusters, **ajouter plus de groupes n'apporte pas beaucoup de valeur supplémentaire**.  

---

### **2. Décision sur le nombre de clusters**  
- **k = 3** : Bonne segmentation générale, plus simple à interpréter.  
- **k = 4** : Permet une **meilleure séparation des comportements financiers** tout en gardant une bonne interprétabilité.  

### **On choisit donc** :  
- **k = 4** pour une **meilleure granularité** dans l'analyse des segments.  



**Application de K-means avec k = 4**

```python
# Application de K-means avec k=4
kmeans = KMeans(n_clusters=4, random_state=42)
df_pca['Cluster'] = kmeans.fit_predict(X)

# Centres des clusters
centroids = kmeans.cluster_centers_
print("Centres des clusters :\n", centroids)
```

# **Application de SVM pour la Classification du Risque de Crédit**

###**Préparation des données pour SVM**

On utilise les composantes principales et les clusters (k=3) comme features.

```python
from sklearn.model_selection import train_test_split

# Features pour SVM
features = ['PC_BILL1', 'PC_BILL2', 'PC_PAY1', 'PC_PAY2', 'Cluster_3']

# Variables indépendantes et cible
X_svm = df_pca[features]
y_svm = df['default.payment.next.month']

# Séparation des données en train/test
X_train, X_test, y_train, y_test = train_test_split(X_svm, y_svm, test_size=0.3, random_state=42)
```

 ### **Application de SVM avec GridSearchCV**

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV

# Features pour SVM
#features = ['PC_BILL1', 'PC_BILL2', 'PC_PAY1', 'PC_PAY2', 'Cluster_3']

# Variables indépendantes et cible
#X_svm = df_pca[features]
#y_svm = df['default.payment.next.month']

# Séparation des données en train/test
#X_train, X_test, y_train, y_test = train_test_split(X_svm, y_svm, test_size=0.3, random_state=42)

# Paramètres à tester
param_grid = {
    'C': [0.1, 1, 10],
    'gamma': ['scale', 'auto'],
    'kernel': ['rbf', 'linear']
}

# GridSearchCV pour trouver les meilleurs paramètres
grid = GridSearchCV(SVC(), param_grid, refit=True, verbose=2, cv=5)
grid.fit(X_train, y_train)

# Meilleurs paramètres
print("Meilleurs paramètres SVM :", grid.best_params_)
```

**Évaluation des performances du modèle**

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Modèle SVM optimisé
best_svm = grid.best_estimator_

# Prédictions sur le set de test
y_pred = best_svm.predict(X_test)

# Évaluation des performances
print("Confusion Matrix :\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report :\n", classification_report(y_test, y_pred))
print("\nAccuracy :", accuracy_score(y_test, y_pred))
```

# **Amélioration du Modèle SVM : Gestion du Déséquilibre des Classes**

Utilisation de SMOTE pour Oversampling

SMOTE (Synthetic Minority Oversampling Technique) permet de :

Créer des exemples synthétiques pour la classe minoritaire (Défaut).

Équilibrer le dataset pour améliorer le rappel de la classe 1.



```python
from imblearn.over_sampling import SMOTE

# Application de SMOTE
sm = SMOTE(random_state=42)
X_resampled, y_resampled = sm.fit_resample(X_train, y_train)

# Vérification de la répartition des classes
print("Répartition après SMOTE :\n", y_resampled.value_counts())
```

**Application de SVM avec Class Weights**

Pour compenser le déséquilibre restant, on utilise class_weight='balanced' dans SVM.

```python
# Modèle SVM avec Class Weights et Paramètres Optimisés
svm_balanced = SVC(class_weight='balanced', C=10, gamma='scale', kernel='rbf', random_state=42)
svm_balanced.fit(X_resampled, y_resampled)

# Prédictions sur le set de test
y_pred_balanced = svm_balanced.predict(X_test)
```

**Évaluation des Performances**


```python
# Évaluation des performances
print("Confusion Matrix (Balanced) :\n", confusion_matrix(y_test, y_pred_balanced))
print("\nClassification Report (Balanced) :\n", classification_report(y_test, y_pred_balanced))
print("\nAccuracy (Balanced) :", accuracy_score(y_test, y_pred_balanced))
```

**Optimisation Avancée des Hyperparamètres avec RandomizedSearchCV**

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform

# Paramètres à explorer
param_distributions = {
    'C': [0.1, 1, 10, 100, 1000],
    'gamma': ['scale', 'auto', 0.001, 0.01, 0.1, 1, 10],
    'kernel': ['rbf', 'linear', 'sigmoid', 'poly'],
    'class_weight': ['balanced', None]
}

# RandomizedSearchCV pour trouver les meilleurs paramètres
random_search = RandomizedSearchCV(SVC(), param_distributions,
                                   n_iter=30, # Nombre d'itérations
                                   scoring='f1', # Optimisation du F1-Score
                                   refit=True,
                                   verbose=2,
                                   cv=5,
                                   random_state=42)
random_search.fit(X_resampled, y_resampled)

# Meilleurs paramètres
print("Meilleurs paramètres SVM (RandomizedSearchCV) :", random_search.best_params_)
```
